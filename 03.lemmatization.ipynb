# Lemmatization

Lemmatization is a text normalization technique in NLP that reduces inflected (and sometimes derivationally related) word forms to their dictionary base form, called a lemma. Unlike stemming, which just chops off affixes, lemmatization uses a vocabulary and morphological analysis (often with part-of-speech information) to ensure the output is a valid word with the correct meaning.


it maps different forms of a word to a single, meaningful base form.

It is slower than Stemming, but more accurate

NLTK provides WordNetLemmatizer class which is a thin wrapper around the wordnet corpus. This class uses morphy() function to the WordNet CorpusReader class to find a lemma.

```python
words = ["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
import nltk
nltk.download("wordnet")
```

```python
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
```

We have to give a value for pos: 

Noun-n

verb-v

adjective-a

adverb-r

Default value of pos: n(noun)

```python
for word in words:
    print(f"Original: {word} | Lemmatized: {lemmatizer.lemmatize(word,pos='v')}")
```

We will get different output lemmas for different values of pos:

```python
for word in words:
    print(f"Original: {word} | Lemmatized: {lemmatizer.lemmatize(word,pos='a')}")
```

```python
lemmatizer.lemmatize("eating", pos='v'), lemmatizer.lemmatize("eats", pos='v'), lemmatizer.lemmatize("eaten", pos='v')
```

```python
lemmatizer.lemmatize("congratulations", pos='n'), lemmatizer.lemmatize("congratulation", pos='n'), lemmatizer.lemmatize("congratulated", pos='v')
```

```python
lemmatizer.lemmatize("goes",pos='v')
```

```python
lemmatizer.lemmatize("fairly",pos='v'),lemmatizer.lemmatize("sportingly")
```

Lemmatization perform much better conversion into root word than Stemming.
