# Stemming 

Stemming is the process of reducing a word to its word stem that affixes to suffixes and prefixes or to the roots of words known as a lemma. Stemming is important in natural language understanding (NLU) and natural language processing (NLP).

Lets take some words to understand

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

## PorterStemmer

A rule-based method that strips common English suffixes in a sequence of steps to produce a compact “stem” for each word. It’s fast, doesn’t need a dictionary.

```python
from nltk.stem import PorterStemmer
porter_stemmer = PorterStemmer()
```

```python
for word in words:
    print(f"Original: {word} | Stemmed: {porter_stemmer.stem(word)}")
```

```python
porter_stemmer.stem("running")
```

```python
porter_stemmer.stem("computer")
```

```python
porter_stemmer.stem("congratulations")
```

```python
porter_stemmer.stem("sitting")
```

From all the examples, it is clear the PorterStemmer is not very accurate, it over stems or under stems some words like congratulations

## RegexpStemmer

A rule-based stemmer that uses a single regular expression to strip affixes (usually suffixes) from words. Instead of a complex multi-step algorithm like Porter, it just removes any substring that matches your regex pattern, optionally only for words longer than a given minimum length

```python
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$|s$|e$|able$', min=4)
```

RegexpStemmer give us the absolute power to set up a regex to stem the word

```python
reg_stemmer.stem('eating')
```

```python
for word in words:
    print(f"Original: {word} | RegexpStemmed: {reg_stemmer.stem(word)}")
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer2 = RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer2.stem('ingeating')
```

## SnowballStemmer

It is a stemming algorithm which is also known as the Porter2 stemming algorithm as it is a better version of the Porter Stemmer since some issues of it were fixed in this stemmer.

Snowball was designed to be faster and more accurate than PorterStemmer and is also multilingual(15+ languages)

```python
from nltk.stem import SnowballStemmer
snowball_stemmer = SnowballStemmer(language="english")
```

```python
for word in words:
    print(f"Original: {word} | Stemmed: {snowball_stemmer.stem(word)}")
```

```python
porter_stemmer.stem("fairly"), porter_stemmer.stem("sportingly")
```

```python
snowball_stemmer.stem("fairly"), snowball_stemmer.stem("sportingly")
```

```python
snowball_stemmer.stem('goes')
```

SnowballStemmer is more accurate than PorterStemmer but it still make mistakes;
That the reason why we use Lemmetization which uses a dictionary based conversion.
