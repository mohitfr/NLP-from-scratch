# N-grams

An n-gram is a contiguous sequence of n items from text or speech; in NLP, the items are usually words/tokens, but they can also be characters. N-grams let you represent short local context that plain Bag of Words loses—for example, distinguishing “not good” from “good.”

## Types of n-grams
**Consider this token sequence:**

Text - **I am studying natural language processing**

Unigram	- 1	-- I, am, studying, natural, language, processing

Bigram  - 2 -- I am, am studying, studying natural, natural language, language processing

Trigram - 3 -- I am studying, am studying natural, studying natural language, natural language processing

4-gram  - 4 -- I am studying natural, am studying natural language, studying natural language processing

- The word contiguous matters: I studying is not a bigram here because am lies between the two words.

- For a sequence of L tokens, the number of n-grams without padding is:

    - **L − n + 1**

    - So a 6-token sentence has: 6 − 2 + 1 = 5 bigrams and 6 − 3 + 1 = 4 trigrams.

## Why n-grams matter
A unigram model only knows which individual words occur. Bigram and trigram features capture short phrases and local word order.

- Example:

    - Text
        Review A: This movie is good

        Review B: This movie is not good

        A unigram representation sees good in both reviews and may struggle with the negation. A bigram representation includes not good, which gives a classifier a much more informative feature.

## Common uses include:

- **Text classification** and **Sentiment analysis**: 
    - Phrases such as not good, highly recommend, or waste of money can be useful features.

- **Language modeling** / **next-word prediction**: 
    - Estimate the next word from the preceding n−1 words.

- **Autocomplete and predictive typing**: 
    - Estimate likely continuations from recent context.

- **Spell checking** and **grammar correction**: 
    - Identify low-probability word sequences.

- **Speech recognition** and **Machine translation**: 
    - Help select plausible word sequences.

- **Search**, **IR**, and **Plagiarism detection**: 
    - Phrase matching and sequence overlap are more informative than individual word matching 

## N-grams in language models
An n-gram language model uses the **Markov assumption**: the probability of the next word depends only on the previous n−1 words.

- For a trigram model: 𝑃(𝑤𝑡 ∣ 𝑤1,...,𝑤𝑡−1) ≈ 𝑃(𝑤𝑡 ∣ 𝑤𝑡−2, 𝑤𝑡−1)

- For example, after natural language, a trigram model could learn that processing is likely: 𝑃(processing ∣ natural, language)

- The whole-sentence probability can then be approximated as a product of these local conditional probabilities.

## N-grams with NLTK

ngrams() function is the direct tool for generating them. It takes a sequence and returns an iterator of tuples

```python
from nltk.util import ngrams
from nltk.tokenize import word_tokenize
```

```python
text = 'I am a student of NLP from scratch. I am learning n-grams in NLP.'
```

```python
tokens = word_tokenize(text.lower())
tokens = [token for token in tokens if token.isalpha()]
```

```python
unigrams = list(ngrams(tokens, 1))
bigrams = list(ngrams(tokens, 2))
trigrams = list(ngrams(tokens, 3))
```

```python
print("Tokens:\n", tokens,end="\n\n")
print("Unigrams:\n", unigrams, end="\n\n")
print("Bigrams:\n", bigrams, end="\n\n")
print("Trigrams:\n", trigrams, end="\n\n")
```

### If we want string phrases instead of tuples:

```python
bigram_phrases = list(" ".join(ngram) for ngram in bigrams)
```

```python
print(f"Bigram Phrases:\n{chr(10).join(bigram_phrases)}")
```

### Padding with sentence markers
For language modeling, it is common to add boundary markers so the model can learn what starts and ends sentences. NLTK supports this through pad_left and pad_right.

```python
tokens = ["i", "love", "nlp"]
```

```python
bigrams = list(
    " ".join(ngram) for ngram in ngrams(
        tokens,
        2,
        pad_left=True,
        pad_right=True,
        left_pad_symbol="<s>",
        right_pad_symbol="</s>"
    )
)

print(bigrams)
```

This is useful because ('< s >', 'i') models a sentence beginning with i, while ('nlp', '</ s >') models a sentence ending after nlp.

## Advantages
- Captures local word order, unlike unigram-only Bag of Words.

- Simple, interpretable, and easy to debug.

- Fast for small n, particularly unigram and bigram features.

- A strong classical baseline for text classification and language modeling.

- Can capture meaningful short phrases such as machine learning, New Delhi, and not bad.

## Limitations
- Vocabulary / feature explosion: with vocabulary size ∣V∣, there can be up to ∣V∣^n possible n-grams, so the space grows rapidly as n increases. 

- Data sparsity: long sequences occur rarely, so many valid n-grams are unseen in training data.

- Limited context: a bigram sees only one previous token; a trigram sees only two, so it misses long-distance dependencies.

- No real semantic understanding: car and automobile remain unrelated features, and the method cannot infer meaning beyond observed local patterns.

- Language models built from counts need smoothing (such as Laplace, Good–Turing, or Kneser–Ney) to avoid assigning zero probability to unseen n-grams.

## Practical rule of thumb
- Start with unigrams + bigrams for a classical text-classification baseline.

- Add trigrams only if you have enough data and a reason to capture short three-word phrases.

- Use frequency thresholds, such as min_df, to remove rare n-grams and control feature growth.

- For modern deep-learning NLP, n-grams remain useful for analysis and baselines, while embedding- and transformer-based models usually handle longer context more effectively.
