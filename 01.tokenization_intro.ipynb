```python
corpus = """This is a sample text for tokenization.
This is python's NLTK library for natural language processing.
It contains multiple sentences!
Tokenization is the process of breaking down text into smaller units called tokens.
"""
```

```python
print("Original Corpus:\n")
print(corpus)
```

## Sentence-->paragraphs

```python
import nltk
nltk.download("punkt_tab")
```

```python
from nltk.tokenize import sent_tokenize
```

```python
documents = sent_tokenize(corpus)

print(documents)

print("\nTokenized Sentences:")
for i, sentence in enumerate(documents):
    print(f"Sentence {i+1}: {sentence}")
```

```python
type(documents)
```

## Paragraph-->words

```python
from nltk.tokenize import word_tokenize
```

```python
words = word_tokenize(corpus)

print(words)

print("\nTokenized Words:")
for i, word in enumerate(words):
    print(f"Word {i+1}: {word}")
```

```python
for sentence in documents:
    print(word_tokenize(sentence))
```

word_tokenize does not seperate punctuation, we can use wordpunct_tokenize

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
words_punct = wordpunct_tokenize(corpus)

for i, word in enumerate(words_punct): 
    print(f"Word {i+1}: {word}")
```

There is another module similar to word_tokenize called TreebankWordTokenize which do not seperate fullstops except the last one:

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer = TreebankWordTokenizer()
words_treebank = tokenizer.tokenize(corpus)
```

```python
for i, word in enumerate(words_treebank):
    print(f"Word {i+1}: {word}")
```
