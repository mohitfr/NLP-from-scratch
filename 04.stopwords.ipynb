# Stopwords in NLP

Stop words are very common words in a language (like “the”, “is”, “and”, “in”) that appear frequently but carry little discriminative meaning for many NLP tasks. In traditional pipelines, they’re often filtered out during preprocessing so that models and indexes focus on the more informative, content-bearing words.

## What stop words are?

Typical stop words include:

Articles: “a”, “an”, “the”

Prepositions: “in”, “on”, “at”, “by”, “for”, “of”

Conjunctions: “and”, “or”, “but”

Common pronouns and auxiliaries: “i”, “you”, “he”, “she”, “it”, “is”, “are”, “was”, “have”, “has”

## They’re characterized by:

High frequency across documents

Low semantic specificity (they don’t help much in distinguishing topics or documents)

## Why stop words are removed
Removing stop words is a common preprocessing step because it:

Reduces noise – eliminates words that contribute little to meaning in many tasks.

Shrinks vocabulary size – fewer unique tokens, which reduces dimensionality for bag-of-words, TF–IDF, and similar representations.

Improves efficiency – less data to store, index, and process; faster training and inference.

Can improve statistical significance – by increasing the signal-to-noise ratio for the remaining terms.

```python
## Speech Of DR APJ Abdul Kalam
corpus = """I have three visions for India. In 3000 years of our history, people from all over 
            the world have come and invaded us, captured our lands, conquered our minds. 
            From Alexander onwards, the Greeks, the Turks, the Moguls, the Portuguese, the British,
            the French, the Dutch, all of them came and looted us, took over what was ours. 
            Yet we have not done this to any other nation. We have not conquered anyone. 
            We have not grabbed their land, their culture, 
            their history and tried to enforce our way of life on them. 
            Why? Because we respect the freedom of others.That is why my 
            first vision is that of freedom. I believe that India got its first vision of 
            this in 1857, when we started the War of Independence. It is this freedom that
            we must protect and nurture and build on. If we are not free, no one will respect us.
            My second vision for India’s development. For fifty years we have been a developing nation.
            It is time we see ourselves as a developed nation. We are among the top 5 nations of the world
            in terms of GDP. We have a 10 percent growth rate in most areas. Our poverty levels are falling.
            Our achievements are being globally recognised today. Yet we lack the self-confidence to
            see ourselves as a developed nation, self-reliant and self-assured. Isn’t this incorrect?
            I have a third vision. India must stand up to the world. Because I believe that unless India 
            stands up to the world, no one will respect us. Only strength respects strength. We must be 
            strong not only as a military power but also as an economic power. Both must go hand-in-hand. 
            My good fortune was to have worked with three great minds. Dr. Vikram Sarabhai of the Dept. of 
            space, Professor Satish Dhawan, who succeeded him and Dr. Brahm Prakash, father of nuclear material.
            I was lucky to have worked with all three of them closely and consider this the great opportunity of my life. 
            I see four milestones in my career"""
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.stem import PorterStemmer, SnowballStemmer, WordNetLemmatizer
from nltk.corpus import stopwords
import time
```

```python
porter_stemmer = PorterStemmer()   
snowball_stemmer = SnowballStemmer("english")
lemmatizer = WordNetLemmatizer()
```

```python
sentences = sent_tokenize(corpus)
```

```python
type(sentences)
```

```python
sentences[:5]
```

Now we need to write a loop on sentences to check if a word in a sentence exist in the set of stop words, if not, then only the word will be stemmed/lemmatized.

Then replace the sentence with joining these new words

We will see the results for Porter Stemmer, Snowball Stemmer and Wordnet Lemmatizer

```python
start_time = time.time()
for i in range(len(sentences)):
    sentences[i] = sentences[i].lower()# converting all the sentences into lower case
    words=word_tokenize(sentences[i])
    words=[porter_stemmer.stem(word) for word in words if word not in stop_words]
    sentences[i]=' '.join(words)# converting all the list of words into sentences
end_time = time.time()
print(f"Time taken: {end_time - start_time} seconds")
```

```python
sentences_snowball = sent_tokenize(corpus)
```

```python
start_time = time.time()
for i in range(len(sentences_snowball)):
    sentences_snowball[i] = sentences_snowball[i].lower()# converting all the sentences into lower case
    words=word_tokenize(sentences_snowball[i])
    words=[snowball_stemmer.stem(word) for word in words if word not in stop_words]
    sentences_snowball[i]=' '.join(words)# converting all the list of words into sentences
end_time = time.time()
print(f"Time taken: {end_time - start_time} seconds")
```

```python
sentences_lemma = sent_tokenize(corpus)
```

```python
start_time = time.time()
for i in range(len(sentences_lemma)):
    sentences_lemma[i] = sentences_lemma[i].lower()# converting all the sentences into lower case
    words=word_tokenize(sentences_lemma[i])
    words=[lemmatizer.lemmatize(word,pos='v') for word in words if word not in stop_words]
    sentences_lemma[i]=' '.join(words)# converting all the list of words into sentences
end_time = time.time()
print(f"Time taken: {end_time - start_time} seconds")
```
