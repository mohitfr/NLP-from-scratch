# Part Of Speech Tagging in NLTK

Part-of-speech (POS) tagging is the NLP task of assigning each word (token) in a sentence a grammatical category like noun, verb, adjective, adverb, etc., based on its role and context. It turns raw text into a structured sequence of (word, tag) pairs, which downstream tasks like parsing, named entity recognition, and semantic analysis rely on.

## What POS tags represent

A POS tag encodes the syntactic category and sometimes finer-grained information (tense, number, degree). Common coarse categories:

Nouns: person, place, thing, idea (e.g., “dog”, “city”, “freedom”)

Verbs: actions or states (e.g., “run”, “is”, “have”)

Adjectives: modify nouns (e.g., “quick”, “brown”)

Adverbs: modify verbs, adjectives, or other adverbs (e.g., “quickly”, “very”)

Pronouns: stand in for nouns (e.g., “he”, “they”)

Prepositions, conjunctions, determiners, etc.

## Why POS tagging matters
### POS tags help models:

**Disambiguate words** with multiple roles:

-“book” as noun vs verb (“I read a book” vs “I will book a ticket”)

-“run” as noun vs verb (“a run” vs “to run”)

Improve downstream tasks:

**Syntactic parsing** and dependency analysis need POS to build correct trees.

**Named entity recognition** uses POS to distinguish proper nouns and modifiers.

**Lemmatization** often requires POS to pick the right lemma (e.g., “meeting” noun vs verb).

**Information extraction**, question answering, and machine translation benefit from knowing word roles.

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer = WordNetLemmatizer()
```

```python
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('punkt')
```

```python
sentence = "The quick brown fox jumps over the lazy dog"
tokens = word_tokenize(sentence)
```

```python
tokens
```

```python
tagged = nltk.pos_tag(tokens)
print(tagged)
```

NLTK’s default pos_tag() uses a pre-trained statistical tagger (based on the Penn Treebank).

```python
corpus = """Biodiversity supports ecosystem resilience and stability.
Cybersecurity protects systems from malicious attacks.
DNA carries genetic information in living organisms.
The sun rises in the east and sets in the west.
Transfer learning leverages pre-trained models.
Reinforcement learning uses rewards to guide agents.
Data structures organize information for efficient access.
The weather today is sunny with mild temperatures.
Convolutional neural networks excel at image recognition.
Virtual reality creates immersive digital experiences."""
```

```python
sentences = nltk.sent_tokenize(corpus)
sentences[:5]
```

```python
for i in range(len(sentences)):
    tokens = word_tokenize(sentences[i])
    tagged = nltk.pos_tag(tokens)
    print(f"Sentence {i+1}: {tagged}")
```

## You can then use these tags for:

Filtering by POS (e.g., only nouns and adjectives for keyword extraction)

Feeding into a lemmatizer with POS-aware rules

Building features for classifiers (e.g., count of verbs, adjectives, etc.)

```python
tagged
```

```python
words=["eating","eats","eaten","writing","writes","programming","programs","history","finally","finalized"]
```

```python
tagged = nltk.pos_tag(words)
tagged
```

```python
# filter words by POS tag
noun_words = [word for word, pos in tagged if pos.startswith('NN')]
print("Noun words:", noun_words)

# filter words by POS tag
verb_words = [word for word, pos in tagged if pos.startswith('VB')]
print("Verb words:", verb_words)
```

```python
## lemmatization with help of pos tags
tagged_with_lemmas = [(word, pos, lemmatizer.lemmatize(word, pos='n' if pos.startswith('NN') else 'v')) for word, pos in tagged]
print("Tagged with lemmas:", tagged_with_lemmas)
```

```python
## lemmatizing specific words with pos tags
lemmatized_specific_pos_noun = [(word, lemmatizer.lemmatize(word, pos='n')) for word, pos in tagged if pos.startswith('NN')]
```

```python
print(lemmatized_specific_pos_noun)
```

## POS tagging with stop words

We can combine POS tagging with stop-word removal, but order matters:

If you remove stop words first, you may lose context that helps the tagger.

Often it’s better to tag first, then filter.

```python
from nltk.corpus import stopwords
```

```python
sentence = "The runners were running and agreed happily."
words = word_tokenize(sentence)
```

```python
tagged = nltk.pos_tag(words)
```

```python
filtered_words = [(word,pos) for word, pos in tagged if word.lower() not in set(stopwords.words('english'))]
```

```python
print("Tagged words (without stopwords):", filtered_words)
```

For some deep-learning-only pipelines (e.g., fine-tuning a transformer end-to-end), you might skip explicit POS tagging because the model learns implicit syntactic information. But for classical ML, feature engineering, or interpretability, POS tags are still very useful.
