# Word Embedding

- Word embeddings are dense, continuous numeric vectors that represent words so that semantic and syntactic relationships are reflected by their positions in a vector space. Words used in similar contexts tend to receive similar vectors.

- They are a major step beyond One-Hot Encoding, Bag of Words, and TF–IDF: rather than treating every word as completely unrelated, embeddings can encode that words like king, queen, man, woman, car, and automobile have useful relationships.

## Core idea
- A word embedding maps each vocabulary word w to a fixed-size dense vector: w -> v(w) ∈ R(d);

    - where d is the embedding dimension—for example, 50, 100, 200, or 300. Static embeddings are commonly represented with roughly 50–300 dimensions, much smaller than a one-hot vector whose length equals the whole vocabulary size.

- For example, given a vocabulary of 10,000 words:

    - One-hot vector for cat: length 10,000, with one 1 and 9,999 zeros.

    - Embedding for cat: perhaps a dense 100-dimensional vector:

    > cat → [0.21, -0.34, 0.08, 0.71, ..., -0.12]

    - The numbers themselves have no standalone human-readable meaning. What matters is the relative geometry: vector similarity, direction, and distance.

## Why embeddings are useful
- The intuition comes from distributional semantics:

    - Words that occur in similar contexts tend to have related meanings.

- For example, these may occur in similar contexts:

    - The cat sat on the mat.
    
    - The dog sat on the mat.

- A model trained on many examples can learn that cat and dog are related, because they often appear around similar words such as the, sat, pet, animal, food, and owner.

- This means embeddings can encode relationships that sparse count-based features cannot:

    - car may be near automobile

    - doctor may be near physician

    - Paris may be near France

    - king - man + woman is often near queen in well-trained classic embeddings

The famous analogy example is useful for intuition, but it is not a guarantee that every embedding model will solve every analogy accurately.

## Embeddings vs previous representations

| Representation   | Unit represented | Vector form             | Captures semantic similarity? | Main limitation                                  |
| ---------------- | ---------------- | ----------------------- | ----------------------------- | ------------------------------------------------ |
| One-Hot Encoding | Individual token | Sparse binary vector    | No                            | Every word is equally unrelated                  |
| Bag of Words     | Document         | Sparse count vector     | No                            | Ignores order and semantic similarity            |
| TF–IDF           | Document         | Sparse weighted vector  | No                            | Weights importance, but words remain independent |
| Word embedding   | Individual token | Dense continuous vector | Yes, to an extent             | Classic embeddings are context-independent       |

- Word embeddings are distributed representations: information about a word is spread over many vector dimensions, rather than stored in one vocabulary index. This makes them much more compact and information-rich than sparse representations.

## How embeddings are learned

- Most classic embeddings are learned from large unlabeled corpora through a self-supervised objective: predict a word from its surroundings, predict surrounding words from a target word, or model word co-occurrence statistics.

### Word2Vec

- Word2Vec is a neural embedding method with two common training architectures:

| Architecture                   | Input              | Prediction target         | Intuition                                       |
| ------------------------------ | ------------------ | ------------------------- | ----------------------------------------------- |
| CBOW (Continuous Bag of Words) | Context words      | Center/target word        | “Given nearby words, predict the missing word.” |
| Skip-gram                      | Center/target word | Surrounding context words | “Given this word, predict its neighbors.”       |

- For example, in: I enjoy learning natural language processing
    - With target learning, nearby words like I, enjoy, natural, and language can form the local context. Word2Vec learns vectors that become useful for predicting these contextual relationships.

### GloVe

- GloVe (Global Vectors for Word Representation) learns vectors from global word–word co-occurrence statistics across the corpus. It uses patterns such as how often ice appears near cold, steam near hot, or doctor near hospital.


### FastText

- FastText extends Word2Vec by representing a word using character n-grams (subwords). For example, a word is decomposed into character pieces, which helps it learn morphology and create a vector for an unseen word from its subword components.

- This makes FastText especially useful for:

    - Morphologically rich languages

    - Misspellings and word variants

    - Out-of-vocabulary words

    - Small datasets where subword sharing can help

## Static vs contextual embeddings

- Classic Word2Vec, GloVe, and FastText embeddings are static: each word has one fixed vector, regardless of context.

- This creates a key problem:
    - I deposited money in the bank.
    - We sat on the bank of the river.

- The word bank has two meanings, but a static embedding assigns it only one vector.

- **Modern transformer models** generate **Contextual embeddings**, meaning the representation for bank changes based on neighboring words. A transformer can produce one vector for financial bank and another for river bank. This is a major reason contextual models such as BERT and GPT outperform static embedding pipelines on many language-understanding tasks.

## Measuring similarity

- A common similarity measure is **Cosine similarity**:

    - cos(θ)= u⋅v / ∥u∥∥v∥

    - ​Value near 1: vectors point in similar directions, often indicating related words.

    - Value near 0: little directional similarity.

    - Value near −1: opposite directions, though interpreting this as “opposite meaning” is not always valid.

- For example, cat and dog may have higher cosine similarity than cat and database, assuming the embeddings were trained on a representative corpus.

## Applications
- Embeddings can be used in:

    - Text classification and sentiment analysis

    - Semantic search and information retrieval

    - Document clustering

    - Named entity recognition

    - Machine translation

    - Question answering

    - Recommendation and similarity systems

    - Sequence models such as RNNs, LSTMs, and transformer-based NLP pipelines

- At the document level, one simple baseline is to average the embedding vectors of all words in the document. That still loses word order, but it can be more semantically meaningful than simple BoW/TF–IDF.

## Advantages

- **Dense and compact**: a 100–300 dimensional vector can replace a huge sparse vocabulary-sized one-hot vector.

- **Captures similarity**: semantically or syntactically related words may be close in embedding space.

- **Reusable**: pretrained embeddings can transfer useful knowledge to downstream NLP tasks.

- **Works well with neural models**: embeddings are trainable inputs to neural architectures.

- **FastText can handle unseen forms**: its character n-gram approach can derive representations for out-of-vocabulary words.

## Limitations
- **Static embeddings ignore context**: standard Word2Vec/GloVe gives a word like bank one vector for all meanings.

- **Training-data bias**: embeddings can learn and amplify social, cultural, and domain biases present in their corpus.

- **Domain mismatch**: an embedding trained on general web text may perform poorly on biomedical, legal, cybersecurity, or code-related language.

- **Out-of-vocabulary issue**: traditional Word2Vec and GloVe cannot directly represent unseen words; FastText reduces, but does not eliminate, this issue.

- **Relationships can be misleading**: closeness means similar distributional usage, not necessarily factual equivalence, truth, or causal relationship.

- **No full sentence understanding**: static token vectors do not by themselves model syntax or long-range dependencies.
